```python
import asyncio

from src.database.session import get_session_factory, init_db
from src.ingestion.pipeline import seed_default_universe
from src.validation.ohlcv_validator import OHLCVCandleValidator

print('Platform ingestion libraries loaded successfully.')
```

```python
validator = OHLCVCandleValidator(min_volume=1.0)

# Sample raw candle verification
valid_candle = {
    'time': 1710000000000,
    'open': 64000.0,
    'high': 64500.0,
    'low': 63800.0,
    'close': 64200.0,
    'volume': 1500.0,
}
report = validator.validate_candle(valid_candle)
print(f'Candle validation status: {report.status}')
print(f'Passed checks: {report.is_valid}')
```

```python
async def inspect_universe():
    await init_db()
    factory = get_session_factory()
    async with factory() as session:
        await seed_default_universe(session)
        print('Database initialized and universe seeded.')

asyncio.run(inspect_universe())
```
