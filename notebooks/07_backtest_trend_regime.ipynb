# 07 - Quantitative Backtesting: Trend Regime Filter Strategy

Backtesting the Trend Regime Filter strategy across bull, bear, and high-volatility chop regimes with dynamic cash de-risking.


```python
import asyncio

from src.backtesting.engine import BacktestConfig, BacktestEngine
from src.database.session import get_session_factory, init_db

print('Trend Regime strategy module ready.')
```

```python
async def run_regime_backtest():
    await init_db()
    config = BacktestConfig(
        strategy_name='TrendRegimeFilter',
        symbols=['BTC', 'ETH', 'SOL'],
        initial_capital=100000.0,
        commission_rate=0.001,
    )
    engine = BacktestEngine(config)
    factory = get_session_factory()
    async with factory() as session:
        res = await engine.run(session)
        print(f'Strategy Total Return: {res.metrics.total_return_pct:.2f}%')
        print(f'Benchmark Return: {res.metrics.benchmark_return_pct:.2f}%')
        print(f'Alpha: {res.metrics.alpha_pct:+.2f}%')

asyncio.run(run_regime_backtest())
```
