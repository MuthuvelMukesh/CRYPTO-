# 02 - Multi-Timeframe Momentum & RVOL Feature Engineering

This notebook computes momentum returns across multiple horizons (1D, 7D, 14D, 30D, 90D), rolling relative volume (RVOL 20-day baseline), and realized volatility using Polars and NumPy.


```python
import numpy as np
import polars as pl

from src.features.momentum import calculate_momentum_features
from src.features.volatility import calculate_volatility_features
from src.features.volume import calculate_volume_features

print('Feature engineering modules imported.')
```

```python
# Generate synthetic price series for demonstration
np.random.seed(42)
n_bars = 200
returns = np.random.normal(0.001, 0.02, n_bars)
prices = 100.0 * np.cumprod(1.0 + returns)
volumes = np.random.exponential(5000, n_bars)

df_market = pl.DataFrame({
    'close': prices,
    'high': prices * 1.01,
    'low': prices * 0.99,
    'volume': volumes,
})
print(f'Generated {len(df_market)} synthetic bars.')
```

```python
mom_feats = calculate_momentum_features(df_market['close'].to_numpy())
vol_feats = calculate_volume_features(df_market['volume'].to_numpy())
volat_feats = calculate_volatility_features(df_market['high'].to_numpy(), df_market['low'].to_numpy(), df_market['close'].to_numpy())

print('Momentum 1D return:', mom_feats.get('return_1d', 0.0))
print('Momentum 7D return:', mom_feats.get('return_7d', 0.0))
print('RVOL 20:', vol_feats.get('rvol_20', 1.0))
print('Realized Volatility 30D:', volat_feats.get('volatility_30d', 0.0))
```
