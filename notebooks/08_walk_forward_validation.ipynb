# 08 - Walk-Forward Validation & Machine Learning Signal Lab

Temporal train/test splitting with embargo buffers to eliminate lookahead bias, out-of-sample prediction evaluations, and Information Coefficient (IC) tracking.


```python
import numpy as np
import pandas as pd

from src.research.ml_models import (
    create_forward_labels,
    walk_forward_train_evaluate,
)

print('Research ML lab initialized.')
```

```python
# Generate sample dataset with momentum and relative strength features
np.random.seed(42)
n = 400
df = pd.DataFrame({
    'close': 100.0 * np.exp(np.cumsum(np.random.normal(0.0005, 0.015, n))),
    'feat_mom_1d': np.random.normal(0, 1, n),
    'feat_rs_btc': np.random.normal(0, 1, n),
    'feat_rvol': np.random.exponential(1.5, n),
})

labeled_df = create_forward_labels(df, horizon_bars=24, price_col='close')
report = walk_forward_train_evaluate(
    labeled_df,
    feature_cols=['feat_mom_1d', 'feat_rs_btc', 'feat_rvol'],
    train_pct=0.70,
    embargo_bars=24,
    model_type='logistic_regression',
)

print(f'Out-of-sample Accuracy: {report.accuracy:.2%}')
print(f'Information Coefficient (IC): {report.information_coefficient:.4f}')
print(f'Spearman Rank IC: {report.rank_ic:.4f}')
print('Learned Feature Weights:', report.feature_importances)
```
