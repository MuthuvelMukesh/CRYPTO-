# 04 - Trend Structure & Macro Market Regime Classification

Identification of macro market regimes: Bull Trend, Bear Trend, High-Volatility Chop, and Low-Volatility Compression using moving average ribbons and trend strength indicators (ADX/MACD).


```python
import numpy as np

from src.features.market_regime import classify_market_regime
from src.features.trend import calculate_trend_features

print('Trend and regime modules loaded.')
```

```python
# Create upward trending market data
n = 250
trend_ret = np.random.normal(0.002, 0.01, n)
closes = 100.0 * np.cumprod(1.0 + trend_ret)
highs = closes * 1.015
lows = closes * 0.985

trend_res = calculate_trend_features(highs, lows, closes)
regime = classify_market_regime(highs, lows, closes)

print('ADX Trend Strength:', trend_res.get('adx', 0.0))
print('EMA 20 > EMA 50:', trend_res.get('ema_20_above_50', False))
print('Macro Regime Classification:', regime.regime.value)
print('Regime Confidence:', regime.confidence)
```
